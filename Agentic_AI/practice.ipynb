```python
import numpy as np

X = np.array([1.5, -0.5], dtype=np.float64)
Y = np.array([1.0, 0.0], dtype=np.float64)
LEARNING_RATE = 0.1

W = np.array([
    [0.2, -0.4],
    [0.7,  0.1]
], dtype=np.float64)

B = np.array([0.1, -0.2], dtype=np.float64)

def forwardDense():
    z = np.dot(W,X)+B
    prediction = 1/(1+np.exp(-z))
    return np.round(z,4) , np.round(prediction,4)

def calculateDenseLoss():
    z,prediction = forwardDense()
    loss = -np.mean(Y*np.log(prediction)+(1-Y)*np.log(1-prediction))
    return float(np.round(loss,4))

def backwardDense():
    z,prediction = forwardDense()
    dz = (prediction-Y)/len(Y)
    dw = np.outer(dz,X)
    db = dz
    return np.round(dw,4) , np.round(db,4)

def trainDenseOneStep():
    z , prediction = forwardDense()
    loss_before = calculateDenseLoss()
    dw , db = backwardDense()
    updated_w = W-LEARNING_RATE*dw
    updated_b = B-LEARNING_RATE*db
    z_new = np.dot(updated_w,X)+updated_b
    pred_new = 1/(1+np.exp(-z_new))

    loss_after = -np.mean(Y*np.log(pred_new)+(1-Y)*np.log(1-pred_new))
    return (
        np.round(updated_w,4),
        np.round(updated_b,4),
        np.round(loss_before,4),
        np.round(loss_after,4)
    )
```

```python
forwardDense()
calculateDenseLoss()
backwardDense()
trainDenseOneStep()
```
